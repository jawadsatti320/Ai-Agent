```python
import cv2
import os
import csv
from datetime import datetime
from deepface import DeepFace

FACES_DB = "/content/faces"
ATTENDANCE_FILE = "/content/attendance/attendance.csv"

# CSV banao
with open(ATTENDANCE_FILE, 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(["Name", "Date", "Time", "Status"])

print("✅ System Ready!")
```

```python
from google.colab import files

# Test photo upload karo
uploaded = files.upload()

for filename in uploaded.keys():
    print(f"\n🔍 Scanning: {filename}")
    try:
        result = DeepFace.find(
            img_path=filename,
            db_path=FACES_DB,
            enforce_detection=False,
            silent=True
        )

        if len(result) > 0 and len(result[0]) > 0:
            match = result[0].iloc[0]['identity']
            name = match.split('/')[-2]

            # Attendance mark karo
            with open(ATTENDANCE_FILE, 'a', newline='') as f:
                writer = csv.writer(f)
                now = datetime.now()
                writer.writerow([name, now.strftime("%Y-%m-%d"),
                                now.strftime("%H:%M:%S"), "Present"])
            print(f"✅ Recognized: {name}")
        else:
            print("❌ Unknown Person!")

    except Exception as e:
        print(f"Error: {e}")
```
